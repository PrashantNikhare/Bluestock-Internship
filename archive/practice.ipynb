```python
Q1 = df["amount_inr"].quantile(0.25)

Q3 = df["amount_inr"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df["amount_inr"] < lower) |
    (df["amount_inr"] > upper)
]

print(outliers)
```
